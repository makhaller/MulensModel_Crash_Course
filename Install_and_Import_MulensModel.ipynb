```python
import matplotlib.pyplot as plt
import MulensModel as mm

# Create a simple point-lens microlensing model
model = mm.Model({
    "t_0": 2452848.06,
    "u_0": 0.133,
    "t_E": 61.5
})

# Plot its magnification over time
model.plot_magnification(
    t_range=[2452810, 2452890],
    subtract_2450000=True
)

plt.xlabel("Time − 2450000")
plt.ylabel("Magnification")
plt.title("Simple Microlensing Model")
plt.show()
```
